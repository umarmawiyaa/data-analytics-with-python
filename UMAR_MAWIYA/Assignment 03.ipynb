Question 1

Write a program that accepts a list from user and print the alternate element of list.

```python
user_list = input("Enter list elements separated by space: ").split()

print("Alternate elements:")
for i in range(0, len(user_list), 2):
    print(user_list[i])
```

Question 2

Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.

```python
user_list = input("Enter list elements separated by space: ").split()

reversed_list = user_list[::-1]

print("Reversed list:", reversed_list)
```

Question 3

Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

Question 4

Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

```python
user_list = input("Enter list elements separated by space: ").split()

rotated_list = [user_list[-1]] + user_list[:-1]

print("Rotated list:", rotated_list)
```

Question 6

Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

```python
date = input("Enter date (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

month_name = months[int(month) - 1]

print(month_name, int(day), year)
```

Question 7

Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized.

```python
def capitalize_words(text):
    words = text.split()
    new_words = []

    for word in words:
        new_words.append(word.capitalize())

    return " ".join(new_words)

text = input("Enter a sentence: ")

print(capitalize_words(text))
```

Question 8

Find the sum of each row of matrix of size m x n.

```python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix = []

for i in range(rows):
    row = list(map(int, input(f"Enter elements of row {i+1}: ").split()))
    matrix.append(row)

for i in range(rows):
    row_sum = sum(matrix[i])
    print(f"Sum of row {i+1} = {row_sum}")
```

Question 9

Write a program to add two matrices of size n x m.

```python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

print("Enter first matrix:")
matrix1 = []

for i in range(rows):
    row = list(map(int, input().split()))
    matrix1.append(row)

print("Enter second matrix:")
matrix2 = []

for i in range(rows):
    row = list(map(int, input().split()))
    matrix2.append(row)

result = []

for i in range(rows):
    row = []
    for j in range(columns):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("Sum of matrices:")
for row in result:
    print(row)
```

Question 10

Write a program to multiply two matrices.

```python
rows1 = int(input("Enter rows of first matrix: "))
cols1 = int(input("Enter columns of first matrix: "))

print("Enter first matrix:")
matrix1 = []

for i in range(rows1):
    row = list(map(int, input().split()))
    matrix1.append(row)

rows2 = int(input("Enter rows of second matrix: "))
cols2 = int(input("Enter columns of second matrix: "))

print("Enter second matrix:")
matrix2 = []

for i in range(rows2):
    row = list(map(int, input().split()))
    matrix2.append(row)

if cols1 != rows2:
    print("Matrix multiplication is not possible.")
else:
    result = [[0 for j in range(cols2)] for i in range(rows1)]

    for i in range(rows1):
        for j in range(cols2):
            for k in range(cols1):
                result[i][j] += matrix1[i][k] * matrix2[k][j]

    print("Multiplication of matrices:")
    for row in result:
        print(row)
```
