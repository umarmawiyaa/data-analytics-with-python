1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd
import numpy as np

data = {
    "name": ["Ali", "Sara", "Ahmed", "Bisma", "Umar", "Ayesha"],
    "age": [25, 32, 28, 35, 40, 30],
    "city": ["Karachi", "Delhi", "Lahore", "Delhi", "Karachi", "Delhi"],
    "salary": [30000, 45000, 35000, 50000, 60000, 42000]
}

df = pd.DataFrame(data)

print("First three rows:")
print(df.head(3))

print("\nShape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
missing_data = {
    "name": ["Ali", "Sara", "Ahmed", "Bisma", "Umar"],
    "age": [25, np.nan, 28, 35, np.nan],
    "salary": [30000, 45000, np.nan, 50000, 60000],
    "city": ["Karachi", "Delhi", "Lahore", np.nan, "Karachi"]
}

df_missing = pd.DataFrame(missing_data)

print("Missing values in each column:")
print(df_missing.isnull().sum())

 
numeric_cols = df_missing.select_dtypes(include="number").columns

df_missing[numeric_cols] = df_missing[numeric_cols].fillna(
    df_missing[numeric_cols].mean()
)
 
df_missing = df_missing.dropna()

print("\nDataFrame after filling and dropping missing values:")
print(df_missing)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
 
sorted_df = df.sort_values(
    by=["salary", "age"],
    ascending=[False, True]
)

print("Sorted DataFrame:")
print(sorted_df)

print("\nTop three highest-paid employees:")
print(df.nlargest(3, "salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
grouped = df.groupby("location").agg(
    average_salary=("salary", "mean"),
    number_of_employees=("name", "count"),
    maximum_salary=("salary", "max")
)

print("Salary statistics by city:")
print(grouped)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame({
    "name": ["Ali", "Sara", "Ahmed", "Bisma"],
    "dept_id": [1, 2, 1, 3],
    "salary": [30000, 45000, 35000, 50000]
})

departments = pd.DataFrame({
    "dept_id": [1, 2, 3],
    "department": ["IT", "HR", "Finance"]
})

 
print("Inner Join:")
print(pd.merge(employees, departments, on="dept_id", how="inner"))

 
print("\nLeft Join:")
print(pd.merge(employees, departments, on="dept_id", how="left"))

 
df1 = employees.copy()

df2 = pd.DataFrame({
    "name": ["Umar", "Sara"],
    "dept_id": [2, 2],
    "salary": [40000, 45000]
})

combined = pd.concat([df1, df2], ignore_index=True)

 
combined = combined.drop_duplicates()

print("\nConcatenated DataFrame without duplicates:")
print(combined)
```
